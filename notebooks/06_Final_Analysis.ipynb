# 06 - Final Analysis

```python
import pandas as pd

df = pd.read_csv("../data/processed/air_quality_cleaned.csv")
pm25 = df[df["pollutant_id"] == "PM2.5"]
pm10 = df[df["pollutant_id"] == "PM10"]
```

```python
# state level results
stations_count = df.groupby("state")["station"].nunique()
avg_pm25 = pm25.groupby("state")["pollutant_avg"].mean()
avg_pm10 = pm10.groupby("state")["pollutant_avg"].mean()

results = pd.DataFrame({
    "total_stations": stations_count,
    "pm25_average": avg_pm25,
    "pm10_average": avg_pm10
})

results = results.round(1).sort_values("pm25_average", ascending=False)
results.to_csv("../outputs/tables/final_results.csv")
results
```

```python
# final summary printouts
print("=== FINAL FINDINGS ===")
print("Total readings analyzed:", len(df))
print("States covered:", df["state"].nunique())
print("Most polluted state (PM2.5):", results.index[0], "with", results["pm25_average"].iloc[0])
print("Least polluted state (PM2.5):", results.dropna().index[-1], "with", results.dropna()["pm25_average"].iloc[-1])

worst_city = pm25.groupby("city")["pollutant_avg"].mean().sort_values(ascending=False).index[0]
print("Most polluted city (PM2.5):", worst_city)
```
